```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()

df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df.head()
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(cancer.data, cancer.target, stratify=cancer.target, random_state=66)
```

```python
from sklearn.neighbors import KNeighborsClassifier
clf = KNeighborsClassifier(n_neighbors=1)
clf.fit(xtrain,ytrain)
```

```python
clf.predict(xtest)
```

```python
clf.score(xtest,ytest)
```

```python
from sklearn.neighbors import KNeighborsClassifier
train_ac = []
test_ac = []
neighbour_setting = range(1,11)
for i in range(1,11):
    clf = KNeighborsClassifier(n_neighbors=i)
    clf.fit(xtrain,ytrain)
    train_ac.append(clf.score(xtrain,ytrain))
    test_ac.append(clf.score(xtest,ytest))
```

```python
import matplotlib.pyplot as plt
plt.plot(neighbour_setting,train_ac, label = "Train accuracy")
plt.plot(neighbour_setting,test_ac, label = "Test accuracy")
```
