```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x= 6* np.random.rand(100,1) - 3
y = 0.5 * x**2 + 1.5*x + 2 + np.random.randn(100,1)
plt.scatter(x,y,color='g')
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(x,y,test_size=0.25, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(xtrain,ytrain)
```

```python
from sklearn.metrics import r2_score

score = r2_score(ytest, lr.predict(xtest))
score
```

```python
plt.plot(xtrain,lr.predict(xtrain), color = 'r' ) 
plt.scatter(xtrain,ytrain)
```

```python
# polynomial tranformatin
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2, include_bias=True)
xtrainpoly = poly.fit_transform(xtrain)
xtestpoly = poly.transform(xtest)
```

```python
xtrainpoly
```

```python
from sklearn.metrics import r2_score

lr = LinearRegression()
lr.fit(xtrainpoly, ytrain)
y_pred = lr.predict(xtestpoly)
score = r2_score(ytest, y_pred)
print(score)
```

```python
plt.scatter(xtrain, lr.predict(xtrainpoly))
plt.scatter(xtrain,ytrain)
```

```python
poly = PolynomialFeatures(degree=3, include_bias=True)
xtrainpoly = poly.fit_transform(xtrain)
xtestpoly = poly.transform(xtest)


lr = LinearRegression()
lr.fit(xtrainpoly, ytrain)
y_pred = lr.predict(xtestpoly)
score = r2_score(ytest, y_pred)
print(score)

plt.scatter(xtrain, lr.predict(xtrainpoly))
plt.scatter(xtrain,ytrain)
```
